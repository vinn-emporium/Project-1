```python
import string as st 
import random 
cursor.execute("use eedc_db")

def Registration():
    try:
        print("\n==== Registration Section ==== ")
        FULL_NAME = input("Enter your Full name: ")
        PHONE_NO = input("Enter your Phone number: ")
        ADDRESS = input("Enter your Address: ")
        STATE = input("Enter state of Purchase: ")
    
        # To get the meter number
        sample = st.digits 
        meter_number = "".join(random.choice(sample) for _ in range(11))
        query = "Insert customers(FULL_NAME, PHONE_NO, ADDRESS, STATE, METER_NUMBER) values(%s, %s, %s, %s, %s)"
        cursor.execute(query,(FULL_NAME,PHONE_NO, ADDRESS,STATE, meter_number))
        database.commit()
        print(f"Registration successful, Your meter number is {meter_number}")
    except Exception as e:
        print(f"Error message: {e}")

def search():
    try:
        FULL_NAME = input("Enter your Full Name:")
        query = "SELECT * from customers where FULL_NAME = %s"
        cursor.execute(query, FULL_NAME)
        result = cursor.fetchone()
        if result is None:
            print("Record not found..")
        else:
            print(f"{result[0]}| {result[1]}| {result[2]}| {result[3]}| {result[4]}| {result[5]}| {result[6]}")
    except Exception as e:
        print(f"Error message: {e}")
        
def update():
    try:
        print("\n=== Update Section ===")
        CUSTOMERS_ID = int(input("Enter Customer id: "))
        FULL_NAME = input("Enter your Full name: ")
        PHONE_NO = input("Enter your Phone number: ")
        ADDRESS = input("Enter your Address: ")
        STATE = input("Enter state of Purchase: ")
    
        query = "update customers set FULL_NAME = %s, PHONE_NO = %s, ADDRESS = %s, STATE = %s where customer_id = %s"
        cursor.execute(query, (FULL_NAME, PHONE_NO, ADDRESS, STATE, CUSTOMERS_ID,))
        database.commit()
        print("Update successful..")
    except Exception as e:
         print(f"Error message: {e}")

def buy_token():
    try: 
        choice = {"TRUE": 1, "FALSE": 0}
        print("\n=== Purchase Token Section ===")
        FULL_NAME = input("Enter your Full name: ")
        METER_NUMBER = input("Enter meter number: ")
        prepaid = input("Is it Prepaid(True/ False):").strip().upper()
        if prepaid not in choice.keys():
            print("Wrong Input..")
        else:
            Amount = float(input("Enter amount: "))
            if Amount <= 0:
                print("wrong inputation")
                
            sample = st.digits
            Tokens = "".join(random.choice(sample) for _ in range(16))
            query = "insert into tokens(FULL_NAME, METER_NUMBER, prepaid, Amount) values (%s, %s, %s, %s)"
            cursor.execute(query, (FULL_NAME, METER_NUMBER, choice[prepaid], Amount))
            database.commit()
            print(f""" ==== Token Details ====
            Full name:{FULL_NAME}, 
            Meter number:{METER_NUMBER},
            prepaid:{prepaid},
            Amount Paid:{Amount},
            Tokens: {Tokens}.   
            """)
            return
    except Exception as e:
         print(f"Error message: {e}")
                   
def display():
    try:
        admin_pin = 2222
        pin = int(input("input Pin: "))
        if pin == admin_pin: 
            query = "SELECT * FROM CUSTOMERS"
            cursor.execute(query)
            result = cursor.fetchall()
            for i in result: 
                print(i)
        else: 
            print("wrong pin")
            return
    except Exception as e:
        print(f"Error message {e}")
        
def delete():
    try:
        pin = 9790
        token = int(input("Input the Token:"))
        if token == pin:
            FULL_NAME = input("Input the full name of the customer: ")
            query = "DELETE  FROM CUSTOMERS WHERE FULL_NAME = %s"
            cursor.execute(query, FULL_NAME)
            database.commit()
            print("Record have been Deleted..")
    except Exception as e:
        print(f"Error message {e}")
        
# Frontend for the system.              
def main():
    while True: 
        print("\n"+ "=" * 50)
        print("==== Welcome Earlycode Electricity Distribution Company ====")
        print("1. >> Registration\n2. >> Confirm\n3. >> Update\n4. >> Purchase Token\n5. >> Display Records\n6. >> Handle Records\n7. >> Logout")
        print("=" * 50)
        option = input("Select an option: ")
        if option == "1":
            Registration()
            continue
        elif option == "2":
            search()
            continue
        elif option == "3":
            update()
            continue
        elif option == "4":
            buy_token()
            continue
        elif option == "5":
            display()
            continue
        elif option == "6":
            delete()
            continue
        elif option == "7":
            print("Thank you for using EEDC...")
            break
        else:
            print("Invaild option..")
if __name__ == "__main__":
    main()
```
